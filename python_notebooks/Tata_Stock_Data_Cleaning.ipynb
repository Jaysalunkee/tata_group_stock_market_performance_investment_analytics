```python
#Converting Data columns
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.info()
```

```python
#Data sorting 
df = df.sort_values(
    by=["Company Name", "Date"]).reset_index(drop=True)
```

```python
#Checking Numerical Columns
df.describe()
```

```python
#checking Missing data
df.isnull().sum()
```

```python
#Checking Duplicates
df.duplicated().sum()
```

```python
df.duplicated(subset=["Company Name", "Date"]).sum()
```

```python
# Check if High price is lower than Open or Close
df[
    (df["High"] < df["Open"]) |
    (df["High"] < df["Close"])
]
```

```python
#Creating Year, Month, and Quarter
df["Year"] = df["Date"].dt.year

df["Month"] = df["Date"].dt.month

df["Month_Name"] = df["Date"].dt.month_name()

df["Quarter"] = "Q" + df["Date"].dt.quarter.astype(str)

df["Year_Month"] = df["Date"].dt.to_period("M").astype(str)
```

```python
#createing price change

df["Price_Change"] = df["Close"] - df["Open"]

df["Price_Change_Percent"] = (
    (df["Close"] - df["Open"]) / df["Open"]
) * 100
```

```python
#Create daily price

df["Price_Range"] = df["High"] - df["Low"]

df["Price_Range_Percent"] = (
    (df["High"] - df["Low"]) / df["Open"]
) * 100
```

```python
df["Trend_Signal"] = np.where(
    df["MA_7"] > df["MA_30"],
    "Bullish",
    np.where(
        df["MA_7"] < df["MA_30"],
        "Bearish",
        "Neutral"
    )
)
```
